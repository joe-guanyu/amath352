```python
import numpy as np
import matplotlib.pyplot as plt

stations = np.array([[0, 0],
                     [60, 10],
                     [20, 50],
                     [70, 45]])
epicenter = np.array([35.03, 25.03])
d = np.array([43.04, 29.12, 29.12, 40.32])

theta = np.linspace(0, 2*np.pi, 200)
labels = ['A', 'B', 'C', 'D']
colors = ['blue', 'orange', 'green', 'purple']

plt.figure(figsize=(9, 7))

for i in range(len(d)):
    x0, y0 = stations[i][0], stations[i][1]

    # Circle around the station
    plt.plot(x0 + d[i]*np.cos(theta),
             y0 + d[i]*np.sin(theta),
             color=colors[i], linewidth=2,
             label='Station ' + labels[i])

    # Station location and label
    plt.plot(x0, y0, 'o', color=colors[i], markersize=7)
    plt.text(x0 + 2, y0 + 2, labels[i], fontsize=12)

plt.plot(epicenter[0], epicenter[1], 'r*',
         markersize=15, label='Epicenter')

plt.xlabel('x (km)')
plt.ylabel('y (km)')
plt.title('Earthquake Epicenter')
plt.axis("equal")
plt.grid(alpha=0.2)
plt.legend(loc='upper left', bbox_to_anchor=(1.02, 1))
plt.tight_layout()
plt.show()
```
